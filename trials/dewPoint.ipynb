```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
```

```python
df = pd.read_csv('../M1_final.csv')
```

```python
print(df['Dew Point'].value_counts().head(30))
```

```python
print(df['Dew Point'].sample(20, random_state=42).to_list())
```

```python
# Step 1: Strip any non-breaking and normal spaces
df['Dew Point'] = (
    df['Dew Point']
    .astype(str)
    .str.replace('\xa0', '', regex=False)
    .str.strip()
)

# Step 2: Convert to numeric, keep real NaNs (don’t fill yet)
df['Dew Point'] = pd.to_numeric(df['Dew Point'], errors='coerce')
```
