# Dataset Analysis for Research Paper

This notebook analyzes the M1_final.csv dataset to extract information needed for the dataset section of the research paper.

```python
# Basic Dataset Information
print("="*60)
print("DATASET OVERVIEW")
print("="*60)
print(f"Dataset Name: M1_final.csv")
print(f"Number of samples (rows): {data.shape[0]:,}")
print(f"Number of features (columns): {data.shape[1]:,}")
print(f"Dataset shape: {data.shape}")
print(f"Memory usage: {data.memory_usage(deep=True).sum() / 1024**2:.2f} MB")
print("="*60)
```

```python
# Feature Analysis
print("\n" + "="*60)
print("FEATURE DESCRIPTIONS")
print("="*60)

# Display column names and data types
print("\nColumns and Data Types:")
for i, (col, dtype) in enumerate(zip(data.columns, data.dtypes), 1):
    non_null_count = data[col].count()
    print(f"{i:2d}. {col:<20} | {str(dtype):<10} | Non-null: {non_null_count:,}")

# Let's examine the first few rows to understand the data better
print("\n" + "="*60)
print("SAMPLE DATA (First 5 rows)")
print("="*60)
data.head()
```

```python
# Identify Target Variable and Feature Categories
print("\n" + "="*60)
print("TARGET VARIABLE IDENTIFICATION")
print("="*60)

# Based on the columns, this appears to be a flight dataset
# Let's identify the likely target variable (probably DEP_DELAY or TAXI_OUT)
potential_targets = ['DEP_DELAY', 'TAXI_OUT']

for target in potential_targets:
    if target in data.columns:
        print(f"\nPotential Target Variable: {target}")
        print(f"Description: {target}")
        print(f"Data type: {data[target].dtype}")
        print(f"Unique values: {data[target].nunique()}")
        print(f"Range: {data[target].min()} to {data[target].max()}")
        print(f"Mean: {data[target].mean():.2f}")
        print(f"Standard Deviation: {data[target].std():.2f}")

# Let's also check the distribution of DEP_DELAY (likely the main target)
if 'DEP_DELAY' in data.columns:
    print(f"\nDEP_DELAY Distribution:")
    print(f"Negative delays (early departures): {(data['DEP_DELAY'] < 0).sum():,}")
    print(f"On-time departures (0 delay): {(data['DEP_DELAY'] == 0).sum():,}")
    print(f"Positive delays (late departures): {(data['DEP_DELAY'] > 0).sum():,}")
    
    # Create delay categories for classification analysis
    data_temp = data.copy()
    data_temp['DELAY_CATEGORY'] = pd.cut(data_temp['DEP_DELAY'], 
                                        bins=[-float('inf'), 0, 15, 60, float('inf')],
                                        labels=['Early/OnTime', 'Short_Delay', 'Medium_Delay', 'Long_Delay'])
    print(f"\nDelay Categories Distribution:")
    print(data_temp['DELAY_CATEGORY'].value_counts())
```

```python
# Missing Values and Data Quality Analysis
print("\n" + "="*60)
print("DATA QUALITY ANALYSIS")
print("="*60)

# Missing values
print("Missing Values per Column:")
missing_values = data.isnull().sum()
missing_percentage = (missing_values / len(data)) * 100

for col in data.columns:
    if missing_values[col] > 0:
        print(f"{col:<20}: {missing_values[col]:,} ({missing_percentage[col]:.2f}%)")

if missing_values.sum() == 0:
    print("No missing values found in the dataset!")
else:
    print(f"\nTotal missing values: {missing_values.sum():,}")

# Check for duplicate records
duplicates = data.duplicated().sum()
print(f"\nDuplicate Records: {duplicates:,}")

# Data completeness
completeness = ((data.count().sum()) / (data.shape[0] * data.shape[1])) * 100
print(f"Data Completeness: {completeness:.2f}%")
```

```python
# Feature Categories and Detailed Descriptions
print("\n" + "="*60)
print("DETAILED FEATURE DESCRIPTIONS")
print("="*60)

# Categorize features
temporal_features = ['MONTH', 'DAY_OF_MONTH', 'DAY_OF_WEEK', 'CRS_DEP_M', 'DEP_TIME_M', 'CRS_ARR_M', 'sch_dep', 'sch_arr']
flight_features = ['OP_UNIQUE_CARRIER', 'TAIL_NUM', 'DEST', 'CRS_ELAPSED_TIME', 'DISTANCE']
weather_features = ['Temperature', 'Dew Point', 'Humidity', 'Wind', 'Wind Speed', 'Wind Gust', 'Pressure', 'Condition']
operational_features = ['DEP_DELAY', 'TAXI_OUT']

print("Feature Categories:")
print(f"1. Temporal Features ({len(temporal_features)}): {', '.join(temporal_features)}")
print(f"2. Flight Features ({len(flight_features)}): {', '.join(flight_features)}")
print(f"3. Weather Features ({len(weather_features)}): {', '.join(weather_features)}")
print(f"4. Operational Features ({len(operational_features)}): {', '.join(operational_features)}")

print(f"\nTotal Features: {len(temporal_features) + len(flight_features) + len(weather_features) + len(operational_features)}")

# Detailed descriptions for research paper
feature_descriptions = {
    'MONTH': 'Month of the flight (1-12)',
    'DAY_OF_MONTH': 'Day of the month (1-31)',
    'DAY_OF_WEEK': 'Day of the week (1-7)',
    'OP_UNIQUE_CARRIER': 'Airline carrier code',
    'TAIL_NUM': 'Aircraft tail number identifier',
    'DEST': 'Destination airport code',
    'DEP_DELAY': 'Departure delay in minutes (negative = early, 0 = on-time, positive = delayed)',
    'CRS_ELAPSED_TIME': 'Scheduled elapsed flight time in minutes',
    'DISTANCE': 'Flight distance in miles',
    'CRS_DEP_M': 'Scheduled departure time in minutes from midnight',
    'DEP_TIME_M': 'Actual departure time in minutes from midnight',
    'CRS_ARR_M': 'Scheduled arrival time in minutes from midnight',
    'Temperature': 'Ambient temperature',
    'Dew Point': 'Dew point temperature',
    'Humidity': 'Relative humidity percentage',
    'Wind': 'Wind direction and characteristics',
    'Wind Speed': 'Wind speed',
    'Wind Gust': 'Wind gust speed',
    'Pressure': 'Atmospheric pressure',
    'Condition': 'Weather condition description',
    'sch_dep': 'Scheduled departure time (encoded)',
    'sch_arr': 'Scheduled arrival time (encoded)',
    'TAXI_OUT': 'Taxi-out time in minutes'
}

print(f"\n{'Feature':<20} | {'Description'}")
print("-" * 80)
for feature, description in feature_descriptions.items():
    print(f"{feature:<20} | {description}")
```

```python
# Temporal Analysis
print("\n" + "="*60)
print("TEMPORAL ANALYSIS")
print("="*60)

# Analyze temporal patterns
print("Month Distribution:")
month_dist = data['MONTH'].value_counts().sort_index()
for month, count in month_dist.items():
    print(f"Month {month:2d}: {count:,} flights ({count/len(data)*100:.1f}%)")

print(f"\nDay of Week Distribution:")
dow_dist = data['DAY_OF_WEEK'].value_counts().sort_index()
for dow, count in dow_dist.items():
    day_names = {1: 'Monday', 2: 'Tuesday', 3: 'Wednesday', 4: 'Thursday', 
                 5: 'Friday', 6: 'Saturday', 7: 'Sunday'}
    day_name = day_names.get(dow, f'Day {dow}')
    print(f"{day_name}: {count:,} flights ({count/len(data)*100:.1f}%)")

# Time period estimation
print(f"\nTime Period Analysis:")
print(f"Data spans across {data['MONTH'].nunique()} unique months")
print(f"Data covers {data['DAY_OF_MONTH'].nunique()} unique days of month")
print(f"Data includes all {data['DAY_OF_WEEK'].nunique()} days of the week")
```

```python
# Categorical Variables Analysis
print("\n" + "="*60)
print("CATEGORICAL VARIABLES ANALYSIS")
print("="*60)

categorical_cols = data.select_dtypes(include=['object']).columns
print(f"Categorical Variables: {list(categorical_cols)}")

for col in categorical_cols:
    unique_count = data[col].nunique()
    print(f"\n{col}:")
    print(f"  Unique values: {unique_count}")
    if unique_count <= 10:  # Show distribution for variables with few categories
        value_counts = data[col].value_counts()
        for value, count in value_counts.head(10).items():
            print(f"    {value}: {count:,} ({count/len(data)*100:.1f}%)")
    else:  # Just show top categories for variables with many categories
        value_counts = data[col].value_counts()
        print(f"  Top 5 categories:")
        for value, count in value_counts.head(5).items():
            print(f"    {value}: {count:,} ({count/len(data)*100:.1f}%)")

# Weather conditions analysis
if 'Condition' in data.columns:
    print(f"\nWeather Conditions Summary:")
    weather_counts = data['Condition'].value_counts()
    print(f"Most common weather: {weather_counts.index[0]} ({weather_counts.iloc[0]:,} flights)")
    print(f"Weather variety: {len(weather_counts)} different conditions")
```
